## Baseline

Includes calculating the mean of the each part in `Train` and apply that to the `Dev`

```python
import os
import shutil
import sys
from glob import glob

import pandas as pd

# torchcodec finds FFmpeg via PATH; a kernel launched before FFmpeg was
# installed inherits a PATH without it, so point at the winget install directly
if shutil.which("ffmpeg") is None:
    _ffmpeg_bins = glob(os.path.join(os.environ["LOCALAPPDATA"], "Microsoft", "WinGet",
                                     "Packages", "Gyan.FFmpeg.Shared*", "*", "bin"))
    if _ffmpeg_bins:
        os.environ["PATH"] = _ffmpeg_bins[0] + os.pathsep + os.environ["PATH"]
```

```python
sys.path.append(os.path.abspath('..'))

from src.metrics import evaluate

df = pd.read_parquet('../data/processed/merged.parquet')
df.shape
```

```python
train_df = df[df['split'] == 'train']
dev_df = df[df['split'] == 'dev']

train_df.shape, dev_df.shape
```

```python
train_df.head()
```

```python
train_means = train_df.groupby('part')["score"].mean()

dev_df['y_true'] = dev_df['score']
dev_df['y_pred'] = dev_df['part'].map(train_means)

dev_df.head()
```

```python
evaluate(dev_df)
```
